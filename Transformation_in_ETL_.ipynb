1.  Define Data Transformation in ETL and explain why it is important.

-  Data Transformation is the T (Transform) phase of the ETL (Extract, Transform, Load) process. It is the process of converting, cleaning, and organizing extracted data into a consistent and usable format before loading it into a target system such as a data warehouse.

Common transformation tasks include:

Cleaning incorrect or missing data
Removing duplicate records
Converting data types and formats (e.g., dates, currencies)
Merging data from multiple sources
Applying business rules and calculations
Standardizing values and naming conventions
Why Data Transformation is Important
Improves Data Quality: Removes errors, duplicates, and inconsistencies to ensure accurate information.
Ensures Consistency: Standardizes data formats, making data from different sources compatible.
Supports Better Decision-Making: Reliable and well-structured data leads to more accurate analysis and business insights.
Enhances Performance: Optimized and organized data improves the speed of reporting and querying.
Meets Business Requirements: Applies business rules so the data aligns with organizational goals and reporting needs.
Facilitates Integration: Makes it easier to combine data from multiple systems into a single repository.

Example:
A company collects customer data from multiple branches. One system stores dates as MM/DD/YYYY, while another uses DD-MM-YYYY. During data transformation, all dates are converted into a single standard format, duplicate customer records are removed, and missing values are handled before the data is loaded into the data warehouse.

In summary, data transformation is essential because it converts raw data into clean, consistent, and meaningful information that is ready for analysis and decision-making.

3.  What is the difference between Normalization and Standardization?

-  | **Normalization**                                    | **Standardization**                                                    |
| ---------------------------------------------------- | ---------------------------------------------------------------------- |
| Scales data to a **fixed range**, usually 0 to 1.    | Transforms data so it has a **mean of 0 and standard deviation of 1**. |
| Useful when the range of values needs to be limited. | Useful when data follows different scales or distributions.            |
| Formula: **(x − min) / (max − min)**                 | Formula: **(x − mean) / standard deviation**                           |
| Example: 50 out of a range 0–100 becomes **0.5**.    | A value equal to the mean becomes **0**.                               |


4.  A dataset has missing values in the “Age” column. Suggest two techniques to handle this and
explain when they should be used.

-  Techniques to Handle Missing Values in the “Age” Column
Mean/Median Imputation
Replace missing age values with the mean or median of the available ages.
Use mean when the data has no major outliers.
Use median when the data contains outliers or is skewed, as the median is less affected by extreme values.
Removing Records
Delete rows where the Age value is missing.
Use this when only a small number of records have missing ages and removing them will not significantly affect the dataset.
Avoid it when many values are missing, as it may lead to loss of important data.

Example: If ages are 20, 21, 22, 23, 80, using the median (22) is better than the mean because 80 is an outlier.

6.  What is One-Hot Encoding? Give an example with the categories: “Red, Blue, Green”.
-  One-Hot Encoding

One-Hot Encoding is a data transformation technique used to convert categorical data into numerical values so that machine learning algorithms can process it.

Each category is represented by a separate column, with:

1 = category is present
0 = category is not present
Example

Categories: Red, Blue, Green

| Color | Red | Blue | Green |
| ----- | --: | ---: | ----: |
| Red   |   1 |    0 |     0 |
| Blue  |   0 |    1 |     0 |
| Green |   0 |    0 |     1 |

So, Red → [1, 0, 0], Blue → [0, 1, 0], and Green → [0, 0, 1].

In simple words: One-Hot Encoding converts categories into 0s and 1s without giving one category a higher numerical value than another.

Min-Max Scaling converts values to a fixed range, usually 0–1. An extreme outlier can stretch the minimum–maximum range and compress most normal values into a very small interval.
Z-score Standardization uses the mean and standard deviation to express values in terms of how far they are from the average. It does not force all values into a fixed range, so normal observations generally retain more useful separation.

Example: If most ages are between 20–50 but one value is 200, Min-Max Scaling can make the 20–50 values appear very close together, while Z-score scaling preserves their relative differences better.
